```python
import polars as pl
import plotly.express as px
from analyses import io
```

```python
celldf = io.read_celldfs("../runs/pairs/", levels=["replica"]).with_columns(
    pl.col("replica").cast(pl.UInt32),
)
celldf
```

```python
clusterdf = celldf.group_by("replica", "wtime").agg(
    cluster_x=pl.col("center_x").mean(),
    cluster_y=pl.col("center_y").mean()
).with_columns(
    displ=(pl.col("cluster_x") ** 2 + pl.col("cluster_y") ** 2) ** 0.5
).sort("replica", "wtime")
clusterdf
```

```python
px.line(clusterdf, x="wtime", y="displ", color="replica")
```

```python
c = 0
for i in range(12, 21, 2):
    c += 20 - i + 1
c
```

```python
px.violin(
    clusterdf,
    y="displ"
).update_layout(
    width=200
)
```

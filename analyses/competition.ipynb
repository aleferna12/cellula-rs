```python
import polars as pl
import plotly.express as px
import plotly.graph_objects as go
import numpy as np
```

```python
grouppers = ["mul_energy", "cell_energy", "mul_n"]
df = pl.read_csv("../comp_results/report.csv").sort(grouppers)
df
```

```python
gdf = df.group_by(grouppers, maintain_order=True).agg(
    pl.col("mul_won").mean(),
    count=pl.col("mul_won").count()
)
gdf
```

```python
px.scatter(
    gdf.filter(pl.col("mul_n") == 15),
    x="mul_energy",
    y="cell_energy",
    color="mul_won"
)
```

```python
gdfm8 = gdf.filter(mul_energy=8)
gdfm12 = gdf.filter(mul_energy=12)
```

```python
am12 = np.reshape(gdfm12["mul_won"].to_numpy(), (4, 3))
am12
```

```python
gdfm12.filter(mul_n=25)
```

```python
px.imshow(
    am12,
    x=gdfm12["mul_n"].unique().cast(str),
    y=gdfm12["cell_energy"].unique().cast(str)
).update_layout(
    xaxis_title="n-mul",
    yaxis_title="cell-energy"
)
```
